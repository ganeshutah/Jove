## 1. The idea


Since $L^0 = \{\varepsilon\}$ for **any** $L$, and $L^* = L^0 \cup \ldots$:

* **$\varepsilon \in L^*$ for every language $L$ whatsoever.**
* In particular $\varepsilon \in \emptyset^*$.
* But $\emptyset^1 = \emptyset$, so $\emptyset^i = \emptyset$ for all $i \ge 1$.
* Therefore **$\emptyset^* = \{\varepsilon\}$**.

Had the Chapter 2 basis been chosen differently, this identity would fail &mdash; and with
it the whole algebra of regular expressions.

## 2. Definitions

### The four-step derivation, executed

```python
print("step 1  L^0 = {epsilon} for any L :", lexp({'ab'},0), lexp(lphi(),0), lexp(lunit(),0))
print("step 2  so epsilon is in every L* :", all('' in lstar(L,3)
      for L in [{'ab'}, lphi(), lunit(), {'0','1'}]))
```

### Steps 3 and 4

```python
print("step 3  empty^i for i>=1 :", [lexp(lphi(), i) for i in range(1,5)])
print("step 4  empty^*          :", lstar(lphi(), 5))
```

## 3. Tests

The headline identity, at several bounds.

```python
for n in range(6):
    assert lstar(lphi(), n) == lunit()
print("lstar(lphi(), n) == {epsilon} for n = 0..5 :", True)
print("value :", lstar(lphi(), 5))
```

$\varepsilon$ is in every star, without exception.

```python
for L, name in [({'ab'},'{ab}'), (lphi(),'{}'), (lunit(),'{eps}'),
                ({'0','1'},'{0,1}'), ({'xyz'},'{xyz}')]:
    print("epsilon in %-8s* ? %s" % (name, '' in lstar(L,3)))
assert all('' in lstar(L,3) for L in [{'ab'}, lphi(), lunit(), {'0','1'}])
```

And $\{\varepsilon\}^* = \{\varepsilon\}$ too &mdash; the other language with a finite star.

```python
print("lstar(lunit(), 5) :", lstar(lunit(), 5))
assert lstar(lunit(),5) == lunit()
print()
print("Exactly TWO languages over any alphabet have a FINITE star:")
print("   {}         -> {epsilon}")
print("   {epsilon}  -> {epsilon}")
print("Anything else contains a non-empty string and so generates infinitely many.")
```
