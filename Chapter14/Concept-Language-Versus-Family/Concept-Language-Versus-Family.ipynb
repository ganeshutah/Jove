## 1. The idea


A persistent confusion: **a language is a point; a family is a region.**

$\Sigma^*$ is a single language. It happens to be regular, so it is **one point**
inside the region labelled "Regular". It is not the region, and it does not contain the
region.

The book's image: imagine spray-painting the plane of all languages. "Regular" is a
small blob; "context-free" a bigger blob containing it; "recursive" bigger still.
Inside the regular blob sit $\emptyset$, $\Sigma^*$, $\{0^n : n \text{ even}\}$ and
infinitely many others &mdash; each a dot.

So $\subseteq$ means two different things depending on the operands, and saying which
one you mean is half of clear thinking here.

## 2. Definitions

### Points (languages) and regions (families)

```python
from itertools import product
def upto(n, sigma='01'):
    return [''.join(p) for k in range(n+1) for p in product(sigma, repeat=k)]

SIGMA_STAR = set(upto(4))
EMPTY      = set()
EVEN0      = {w for w in upto(4) if w.count('0') % 2 == 0}
```

### Deciding regularity of a point, by exhibiting a machine

```python
def is_regular_witness(D, members, upto_n=4):
    return {w for w in upto(upto_n) if accepts_dfa(D, w)} == members
```

## 3. Tests

Three **points**, all inside the **region** 'Regular'.

```python
DSigma = md2mc('DFA\nIF : 0 | 1 -> IF\n')
DEmpty = md2mc('DFA\nI : 0 | 1 -> I\n')
DEven0 = md2mc('''DFA
IF : 0 -> Od
IF : 1 -> IF
Od : 0 -> IF
Od : 1 -> Od
''')
for name, D, L in [('Sigma*', DSigma, SIGMA_STAR), ('empty', DEmpty, EMPTY),
                   ('even 0s', DEven0, EVEN0)]:
    ok = is_regular_witness(D, L)
    print("  %-8s is a regular LANGUAGE (a point)?  %s" % (name, ok))
    assert ok
```

**Containment of languages** is set inclusion between points.

```python
print("empty   subset of Sigma* ?", EMPTY <= SIGMA_STAR)
print("even 0s subset of Sigma* ?", EVEN0 <= SIGMA_STAR)
print("Sigma*  subset of even 0s?", SIGMA_STAR <= EVEN0)
assert EMPTY <= SIGMA_STAR and not SIGMA_STAR <= EVEN0
```

**Containment of families** is inclusion between regions.

```python
FAMILIES = ["regular", "deterministic CFL", "context-free",
            "context-sensitive", "recursive", "recursively enumerable"]
for a, b in zip(FAMILIES, FAMILIES[1:]):
    print("  %-20s  STRICTLY INSIDE  %s" % (a, b))
print("\nEach is a region containing the previous one, with room to spare.")
```

The confusion, stated so you can catch yourself making it.

```python
print("WRONG : 'Sigma* contains the regular languages'")
print("RIGHT : 'Sigma* IS a regular language; it is one point in that family'")
print()
print("WRONG : 'regular is a subset of Sigma*'")
print("RIGHT : 'every regular language is a subset of Sigma*'")
print("        (each POINT in the region is a subset of the POINT Sigma*)")
```

Both readings are true at once for $\Sigma^*$, which is why it trips people.

```python
print("Sigma* is a regular language          -- a point in the region")
print("every regular language is inside it   -- as a SET of strings")
print()
print("Two different relations, both holding.  Say which one you mean.")
```
