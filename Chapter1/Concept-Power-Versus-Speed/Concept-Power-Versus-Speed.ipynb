## 1. The idea


Practically, a supercomputer beats a phone. Formally, **given enough time and memory,
anything solvable on one is solvable on the other**. The same *types* of problems.

Why exclude time? Because it is *ephemeral* &mdash; machines keep getting faster and keep
differing. A theory indexed on wall-clock time would be a theory of hardware.

Cost is not abandoned, only quarantined: **complexity theory** (Chapter 16) studies it.

## 2. Definitions

### Two implementations of one function

Both compute Fibonacci. One is exponential, one linear. **Same outputs, wildly
different costs.**

```python
import time

def fib_slow(n):
    """Exponential: recomputes everything."""
    return n if n < 2 else fib_slow(n-1) + fib_slow(n-2)

def fib_fast(n):
    """Linear: remembers the last two."""
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

### A timing helper

We measure so we can *see* that the gap is enormous &mdash; and that it changes no answer.

```python
def timed(f, n):
    t0 = time.time()
    v  = f(n)
    return v, time.time() - t0
```

## 3. Tests

Identical answers.

```python
same = all(fib_slow(n) == fib_fast(n) for n in range(18))
print("fib_slow and fib_fast agree on 0..17 :", same)
assert same
```

Wildly different times. The slow one is the "cellphone"; the fast one the
"supercomputer". **They compute the same function.**

```python
for n in [18, 22, 26]:
    vs, ts = timed(fib_slow, n)
    vf, tf = timed(fib_fast, n)
    print("n=%2d  value=%-6d  slow=%7.4fs  fast=%.6fs  ratio~%d"
          % (n, vs, ts, tf, int(ts / max(tf, 1e-9))))
```

The boundary this concept draws.
