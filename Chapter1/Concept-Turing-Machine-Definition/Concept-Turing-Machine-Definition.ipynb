## 1. The idea


Turing's recipe, in full:

* read the character under the head,
* write a character,
* move the head **left, right, or stay**,
* go to the next state.

Stop at a **halting state**; whatever is on the tape is the answer.

Nothing about speed appears anywhere. A machine doing three operations a second and a
modern laptop compute exactly the same things.

## 2. Definitions

### A Turing machine that flips every bit

In Jove's markdown, a TM line reads `state : read ; write , move -> next`.
The blank cell is written `.`

```python
flipper = md2mc('''TM
I : 0 ; 1 , R -> I
I : 1 ; 0 , R -> I
I : . ; . , S -> F
''')
print("states :", sorted(flipper["Q"]))
assert flipper["q0"] == "I" and flipper["F"] == {"F"}, "I = initial, F = final"
```

Draw it. The machine is tiny &mdash; that is the point Turing was making.

```python
dotObj_tm(flipper, FuseEdges=True)
```

## 3. Tests

Run it. `explore_tm(machine, tape, fuel)` runs until it halts **or the fuel runs out**.

The `fuel` argument exists because halting is undecidable: a simulator with no
budget could hang forever. A "no result" therefore means *not within this budget*.

```python
explore_tm(flipper, "0101", 60)
```

Try a longer tape, and one that is already all ones.

```python
explore_tm(flipper, "111000", 100)
```
