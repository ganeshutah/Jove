## 1. The idea


Pascal's 1642 calculator could **add and subtract directly, and multiply and divide by
repetition**. Babbage's Analytical Engine (1837) was programmed by Ada Lovelace.
Both were mechanical, and neither was a *computer* in our sense.

The question the 20th century asked was not "is this machine fast?" but
**"what can a machine do at all?"** &mdash; and to ask that precisely, you first have to
say what a machine *is*. That is the job of the rest of this book.

## 2. Definitions

### Pascal's trick: multiplication as repeated addition

Pascal's machine had no multiplier. It multiplied by *repeating* addition &mdash;
which is exactly how a Turing machine will later multiply.

```python
def pascal_multiply(a, b):
    """Multiply using only addition and subtraction, as Pascal's machine did."""
    total, count = 0, 0
    while count < b:
        total = total + a      # the only arithmetic the machine really had
        count = count + 1
    return total

print(pascal_multiply(7, 6), pascal_multiply(12, 12))
```

### Lovelace's insight: a machine can follow a *program*

Ada Lovelace wrote the first algorithm for the Analytical Engine (Bernoulli numbers).
The point was not the numbers: it was that the *sequence of steps* is itself data you
can hand to a machine.

```python
def run_program(program, x):
    """A tiny 'engine': it executes a list of operations given as data."""
    for (op, arg) in program:
        if   op == 'add': x = x + arg
        elif op == 'mul': x = pascal_multiply(x, arg)
        elif op == 'sub': x = x - arg
    return x

square_then_add3 = [('mul', 0), ('add', 0)]   # placeholder, see tests
print(run_program([('add', 5), ('mul', 3), ('sub', 1)], 0))
```

## 3. Tests

Check the mechanical multiplier against Python's own `*`. Same answers, far more steps &mdash;
which is the first hint that **speed and power are different questions**.

```python
ok = all(pascal_multiply(a, b) == a * b
         for a in range(0, 13) for b in range(0, 13))
print("pascal_multiply agrees with * on 0..12 :", ok)
assert ok
```

A program is data: we can *build* one at run time and feed it to the engine.

```python
# "add 1, seven times" -- built programmatically
prog = [('add', 1)] * 7
print("engine result :", run_program(prog, 0))
print("program itself:", prog)
assert run_program(prog, 0) == 7
```
