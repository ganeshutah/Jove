## 1. The idea


Star is officially Chapter 3's, but its digestible form is here:

$$star(L,0)=L^0, \quad star(L,1)=L^0\cup L^1, \quad star(L,2)=L^0\cup L^1\cup L^2$$

Then $star(L) = star(L,\infty)$.

The trap in the chapter's own prompt: "an arbitrary string that is finite/infinite..."
&mdash; the answer must be **finite**. Taking the *bound* to infinity while every *string*
stays finite is the move Chapter 3 formalises.

## 2. Definitions

### Star, bounded &mdash; by hand and via Jove

```python
def star_upto(L, n):
    out = lunit()
    for i in range(1, n+1):
        out = lunion(out, lexp(L, i))
    return out

print("star({0,1},2) :", sorted(lstar({'0','1'}, 2)))
```

### Counting: a geometric sum

```python
def star_size(k, n):
    """|star(Sigma,n)| when |Sigma| = k and nothing collides."""
    return sum(k**i for i in range(n+1))
```

## 3. Tests

The hand version and Jove's `lstar` agree.

```python
for L in [{'0','1'}, {'0','10'}, {'0','1','00',''}]:
    for n in range(4):
        assert star_upto(L,n) == lstar(L,n), (L,n)
print("star_upto matches lstar on every case tested :", True)
```

The book's worked values.

```python
print("star({0,1},2)   :", sorted(lstar({'0','1'},2)))
print("star({0,10},2)  :", sorted(lstar({'0','10'},2)))
print("star({0,1,00,''},2) has", len(lstar({'0','1','00',''},2)), "strings")
assert lstar({'0','1'},2) == {'','0','1','00','01','10','11'}
```

$|star(\{0,1\},n)| = 2^{n+1}-1$ &mdash; and collisions break the formula when they occur.

```python
for n in range(5):
    print("n=%d  actual %-3d  formula 2^%d - 1 = %d"
          % (n, len(lstar({'0','1'},n)), n+1, 2**(n+1)-1))
assert all(len(lstar({'0','1'},n)) == 2**(n+1)-1 for n in range(5))
print()
print("For {0,1,00,''} the formula FAILS -- epsilon and overlaps collide. Check it.")
```

The key point about the limit.

```python
big = lstar({'0','1'}, 7)
print("bound 7 -> %d strings, longest is %d symbols" % (len(big), max(len(s) for s in big)))
print()
print("As n -> infinity the SET becomes infinite,")
print("but every STRING in it stays finite. That is star.")
```
