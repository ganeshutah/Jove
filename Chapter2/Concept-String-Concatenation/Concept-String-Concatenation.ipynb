## 1. The idea


Concatenating $ab$ with $cde$ is written $ab\,cde$ and gives $abcde$. In Python that is
`"ab" + "cde"`.

It is **associative** and **not commutative**. That is exactly the profile of
multiplication, which is why the next concept treats it as one.

## 2. Definitions

### Concatenation, and its two algebraic properties

```python
def cat(x, y):
    return x + y

print(cat('ab', 'cde'))
```

### $\varepsilon$ leaves a string alone

```python
print(repr(cat('', 'ab')), repr(cat('ab', '')))
```

## 3. Tests

Associative: the bracketing does not matter.

```python
x, y, z = 'ab', 'cd', 'ef'
print("(xy)z :", cat(cat(x,y), z))
print("x(yz) :", cat(x, cat(y,z)))
assert cat(cat(x,y), z) == cat(x, cat(y,z))
```

**Not** commutative &mdash; order changes the answer.

```python
print("xy :", cat(x,y), "   yx :", cat(y,x))
assert cat(x,y) != cat(y,x)
print()
print("Associative but not commutative -> behaves like MULTIPLICATION, not addition.")
```

Length adds, even though the strings do not commute.

```python
for a, b in [('ab','cde'), ('', 'xyz'), ('q','')]:
    print("|%s| + |%s| = %d = |%s|" % (a, b, len(a)+len(b), cat(a,b)))
```
