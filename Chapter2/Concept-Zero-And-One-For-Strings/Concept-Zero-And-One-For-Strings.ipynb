## 1. The idea


Whenever you meet an operator resembling multiplication, ask two questions:

* what is its **One** &nbsp;($x \cdot One = x$)?
* what is its **Zero** ($x \cdot Zero = Zero$)?

For numbers: $1$ and $0$. For **string** concatenation: **One is $\varepsilon$** &mdash; and
there is **no Zero at all**.

That missing Zero is what makes the *language*-level answer (Concept 10) surprising.

## 2. Definitions

### The laws, for numbers

```python
def check_num_laws(x):
    return (x * 1 == x, x * 0 == 0)

print("65 :", check_num_laws(65))
print("0.557 :", check_num_laws(0.557))
```

### The laws, for strings

```python
def is_one_for_strings(e, samples):
    return all(e + s == s and s + e == s for s in samples)

def try_zero_for_strings(z, samples):
    """Is there a string z with  z+s == z  and  s+z == z  for all s?"""
    return all(z + s == z and s + z == z for s in samples)
```

## 3. Tests

$\varepsilon$ is the One for string concatenation.

```python
samples = ['', 'a', 'ab', 'hello']
print("epsilon is One? :", is_one_for_strings('', samples))
assert is_one_for_strings('', samples)
```

There is **no Zero**. Try every short candidate and watch them all fail.

```python
cands = ['', 'a', 'b', 'ab', 'zzz']
for z in cands:
    print("candidate %-6r works as Zero? %s" % (z, try_zero_for_strings(z, samples)))
assert not any(try_zero_for_strings(z, samples) for z in cands)
print()
print("No string absorbs every other string. String concatenation has NO Zero.")
```

Keep this gap in mind: at the **language** level a Zero *does* appear.

```python
print("strings   : One = epsilon        Zero = (none)")
print("languages : One = {epsilon}      Zero = {}        <- Concept 10")
```
