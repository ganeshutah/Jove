## 1. The idea


Unlike strings, languages have **both** a Zero and a One.

* **Zero is $\emptyset$**: $\emptyset L = L \emptyset = \emptyset$. You cannot "find an
  $x \in \emptyset$", so no pair forms.
* **One is $\{\varepsilon\}$**: $\{\varepsilon\}L = L\{\varepsilon\} = L$.

Keep three things apart: $\emptyset$ (no strings), $\{\varepsilon\}$ (one string, the empty
one), and $\{\emptyset\}$ (a set containing the empty *set* &mdash; not a language at all).

## 3. Tests

$\emptyset$ absorbs.

```python
L = {'ab', 'c'}
print("lcat(lphi(), L) :", lcat(lphi(), L))
print("lcat(L, lphi()) :", lcat(L, lphi()))
assert lcat(lphi(), L) == lphi() == lcat(L, lphi())
print("  -> Zero absorbs from both sides.")
```

$\{\varepsilon\}$ preserves.

```python
print("lcat(lunit(), L) :", sorted(lcat(lunit(), L)))
print("lcat(L, lunit()) :", sorted(lcat(L, lunit())))
assert lcat(lunit(), L) == L == lcat(L, lunit())
print("  -> One leaves L alone from both sides.")
```

Side by side. **Different sizes, opposite behaviour.**

```python
print("%-12s %-6s %s" % ("language", "size", "L . that"))
print("%-12s %-6d %s" % ("{}",         len(lphi()),  sorted(lcat(L, lphi()))))
print("%-12s %-6d %s" % ("{epsilon}",  len(lunit()), sorted(lcat(L, lunit()))))
print()
print("Do NOT write {} in Python for the empty LANGUAGE -- that is an empty dict!")
print("type({}) is", type({}).__name__, "; use set() or lphi().")
```
