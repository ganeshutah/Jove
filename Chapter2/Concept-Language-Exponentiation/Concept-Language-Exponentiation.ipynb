## 1. The idea


$L^0 = \{\varepsilon\}$, &nbsp; $L^n = L\,L^{n-1}$.

The basis holds for **any** $L$ &mdash; *including the empty language*. Therefore

$$\emptyset^0 = \{\varepsilon\}$$

The usual protest is "but $0^0$ is undefined!" &mdash; and the answer is that we are not
raising the **number** 0 to the **number** 0. We are raising the **language**
$\emptyset$ to the **number** 0. Different operation entirely.

## 2. Definitions

### The definition, and Jove's `lexp`

```python
def lexp_by_hand(L, n):
    return lunit() if n == 0 else lcat(L, lexp_by_hand(L, n-1))

print("lexp({'ab','bc'},2) :", sorted(lexp({'ab','bc'}, 2)))
```

### The operational picture

$L^n$ = "select $n$ strings from $L$, repetitions allowed, and concatenate them".

```python
M = {'0', '10'}
print("M^2 :", sorted(lexp(M,2)), " <- 0+0, 0+10, 10+0, 10+10")
```

## 3. Tests

The headline result, run twice.

```python
print("lexp(lphi(), 0) :", lexp(lphi(), 0), "  <- the ONE language {epsilon}")
print("lexp(lphi(), 1) :", lexp(lphi(), 1), "  <- empty")
print("lexp(lphi(), 5) :", lexp(lphi(), 5), "  <- still empty")
assert lexp(lphi(),0) == lunit()
assert all(lexp(lphi(),i) == lphi() for i in range(1,6))
print()
print("Those two lines together ARE the proof that empty^* = {epsilon} (Ch 3).")
```

An alphabet **is** a language, so $\Sigma^n$ is just exponentiation.

```python
Sigma = {'0','1'}
for n in range(4):
    print("Sigma^%d : %s" % (n, sorted(lexp(Sigma, n))))
assert lexp(Sigma,0) == {''} and len(lexp(Sigma,3)) == 8
```

Jove agrees with the hand-written recursion.

```python
ok = all(lexp(L,n) == lexp_by_hand(L,n)
         for L in [{'a'}, M, Sigma, lphi(), lunit()] for n in range(4))
print("lexp matches the definition everywhere tested :", ok)
assert ok
```
