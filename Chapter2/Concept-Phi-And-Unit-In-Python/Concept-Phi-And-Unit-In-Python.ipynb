## 1. The idea


`lphi()` returns the **Zero** language $\emptyset$; `lunit()` returns the **One**,
$\{\varepsilon\}$.

They are **functions taking no arguments** rather than module constants, so a caller
cannot accidentally mutate a shared set.

The warning that matters: **do not write `{}`** for the empty language. Python reads
that as an empty *dict*.

## 2. Definitions

### The two functions

```python
print("lphi()  ->", repr(lphi()))
print("lunit() ->", repr(lunit()))
```

### Why functions, not constants

```python
a = lphi(); b = lphi()
a.add('oops')                 # mutate one copy
print("a :", a)
print("b :", b, " <- unaffected, because lphi() built a fresh set")
assert b == set()
```

## 3. Tests

The `{}` trap, demonstrated.

```python
print("type({})      :", type({}).__name__,  "  <- a DICT, not a set!")
print("type(set())   :", type(set()).__name__)
print("type(lphi())  :", type(lphi()).__name__)
assert isinstance({}, dict) and isinstance(lphi(), set)
```

They really are the base cases of the recursive operations.

```python
print("lexp(anything, 0)  ->", lexp({'ab'}, 0), " == lunit()?", lexp({'ab'},0) == lunit())
print("lstar(anything, 0) ->", lstar({'ab'}, 0), " == lunit()?", lstar({'ab'},0) == lunit())
assert lexp({'ab'},0) == lunit() == lstar({'ab'},0)
```

A wrong base case corrupts everything downstream, silently.

```python
def bad_lexp(L, n):
    return set() if n == 0 else lcat(L, bad_lexp(L, n-1))   # Zero instead of One

print("good lexp({'ab'},3) :", sorted(lexp({'ab'},3)))
print("bad  lexp({'ab'},3) :", sorted(bad_lexp({'ab'},3)), " <- everything vanished")
```
