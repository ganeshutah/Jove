## 1. The idea


$L_1 \cup L_2$ and $L_1 \cap L_2$ are nothing but set union and intersection. In Jove,
`lunion` is `L1 | L2` and `lint` is `L1 & L2`.

The instructive part is what is **absent**: **complementation is not defined here**.
It needs a *universal set over an alphabet*, and $\Sigma^*$ has not been built yet.

Jove also ships containment tests the chapter does not mention &mdash; `lissubset` and
`lissuperset` &mdash; which are how you actually test a claimed identity.

## 2. Definitions

### The four set operations Jove gives you now

```python
L1 = {'a', 'ab', 'b'}
L2 = {'b', 'ba'}
print("lunion    :", sorted(lunion(L1,L2)))
print("lint      :", sorted(lint(L1,L2)))
print("lminus    :", sorted(lminus(L1,L2)))
print("lsymdiff  :", sorted(lsymdiff(L1,L2)))
```

### The containment tests

```python
print("lissubset(L1,L1)   :", lissubset(L1,L1))
print("lissubset({'a'},L1):", lissubset({'a'}, L1))
print("lissuperset(L1,{'a'}) :", lissuperset(L1, {'a'}))
```

## 3. Tests

They agree with Python's own set operators.

```python
assert lunion(L1,L2) == (L1 | L2)
assert lint(L1,L2)   == (L1 & L2)
assert lminus(L1,L2) == (L1 - L2)
print("lunion/lint/lminus match |, &, - exactly :", True)
```

**Testing a claimed identity** by double containment on a finite approximation.

```python
A = lstar({'0'}, 4)
B = {'0'*i for i in range(5)}
print("A subset of B ?", lissubset(A,B), "   B subset of A ?", lissubset(B,A))
print("=> equal on this window :", lissubset(A,B) and lissubset(B,A))
assert lissubset(A,B) and lissubset(B,A)
```

The missing operation, and why.

```python
print("Jove has lcomplem(L, Sigma, n) -- THREE arguments.")
print("  Sigma : because complement is meaningless without a universe")
print("  n     : because that universe (Sigma*) is infinite and must be bounded")
print()
print("Neither argument existed yet in Chapter 2. Hence the deferral.")
```
