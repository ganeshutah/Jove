## 1. The idea


Infinite languages cannot be printed. The remedy used throughout Jove is **finite
approximation**: show every string under a given bound.

It gives a rough idea without seeing everything &mdash; exactly as $\pi = 3.1415926$ does.

The skill to build here is translating a set-builder definition into a comprehension,
**in both directions**.

## 2. Definitions

### From set-builder to comprehension

```python
# L = { a^i b^j : i,j >= 0 and i < j },  approximated to i,j <= 9
La_lt_b_9 = {"a"*i + "b"*j for i in range(10) for j in range(10) if i < j}
print("size :", len(La_lt_b_9))
```

### A reusable "bounded view" helper

Sets are unordered, so we sort for display &mdash; **order within a set does not matter**.

```python
def view(L, upto=None, n=12):
    xs = sorted(L, key=lambda s: (len(s), s))
    if upto is not None:
        xs = [s for s in xs if len(s) <= upto]
    return xs[:n]
```

## 3. Tests

A window onto the language, shortest first.

```python
print(view(La_lt_b_9, upto=4))
```

Growing the bound reveals more. It never reveals *all* &mdash; there is no all.

```python
for b in [3, 5, 9]:
    Lb = {"a"*i + "b"*j for i in range(b+1) for j in range(b+1) if i < j}
    print("bound %d -> %3d strings, longest %d" % (b, len(Lb), max(len(s) for s in Lb)))
```

Jove's own `lstar` is a finite approximation in exactly this sense.

```python
approx = lstar({'0','1'}, 3)
print("lstar({0,1},3) size :", len(approx))
print("sample              :", view(approx, n=10))
assert '' in approx and '111' in approx
print()
print("This is all strings of length <= 3. The true {0,1}* is infinite.")
```
