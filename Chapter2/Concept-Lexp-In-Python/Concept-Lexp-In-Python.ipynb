## 1. The idea


```python
def lexp(L, n):
    return lunit() if n == 0 else lcat(L, lexp(L, n-1))
```

`lunit()` **is** the $L^0 = \{\varepsilon\}$ basis; `lcat(L, lexp(L,n-1))` **is**
$L\,L^{n-1}$. The correspondence is exact &mdash; the mathematics was designed to be run.

The chapter insists the exercises below be done before moving on.

## 2. Definitions

### Exercising it on $\Sigma$

````python
Sigma = {'0','1'}
for n in range(4):
    print("Sigma^%d (%d strings): %s" % (n, len(lexp(Sigma,n)), sorted(lexp(Sigma,n))))
````

### And on a language where collisions happen

````python
M = {'0','10'}
for n in range(4):
    print("M^%d (%d strings): %s" % (n, len(lexp(M,n)), sorted(lexp(M,n))))
````

## 3. Tests

$|\Sigma^n| = 2^n$ exactly, because no two selections collide.

````python
for n in range(6):
    print("|Sigma^%d| = %-3d  2^%d = %d" % (n, len(lexp(Sigma,n)), n, 2**n))
assert all(len(lexp(Sigma,n)) == 2**n for n in range(6))
````

For $M = \{0,10\}$ the counts fall **below** $2^n$ &mdash; selections collide.

````python
for n in range(5):
    print("|M^%d| = %-3d  vs 2^%d = %d" % (n, len(lexp(M,n)), n, 2**n))
print()
print("Collisions: different selections build the same string, and the set absorbs it.")
````

The mandatory exercise, answered by machine.

````python
print("Sigma^0 :", sorted(lexp(Sigma,0)))
print("Sigma^1 :", sorted(lexp(Sigma,1)))
print("Sigma^2 :", sorted(lexp(Sigma,2)))
print("Sigma^3 :", sorted(lexp(Sigma,3)))
print("M^2     :", sorted(lexp(M,2)), "<- matches the book's worked case")
assert sorted(lexp(M,2)) == ['00','010','100','1010']
````
