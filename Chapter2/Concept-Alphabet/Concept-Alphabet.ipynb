## 1. The idea


An **alphabet** $\Sigma$ is a *finite*, *non-empty* set of symbols.

Both adjectives matter. **Finiteness** is what lets a machine have finitely many
transition rules. **Non-emptiness** rules out degenerate cases.

Convention: pick $\Sigma$ and **stick with it** &mdash; do not change it mid-construction.

## 2. Definitions

### Some alphabets

```python
Sigma_bin = {'0', '1'}
Sigma_ab  = {'a', 'b'}
Sigma_one = {'a'}            # a singleton alphabet
print("binary    :", sorted(Sigma_bin))
print("singleton :", sorted(Sigma_one))
```

### Checking the two requirements

```python
def is_alphabet(S):
    return (isinstance(S, set) and len(S) > 0
            and all(isinstance(x, str) and len(x) == 1 for x in S))
```

## 3. Tests

The empty set is not an alphabet; neither is an infinite one.

```python
for S, why in [({'0','1'}, 'binary'), (set(), 'empty'),
               ({'ab'}, 'multi-char member'), ({'a'}, 'singleton')]:
    print("%-22s alphabet? %s" % (why, is_alphabet(S)))
assert is_alphabet({'0','1'}) and not is_alphabet(set())
```

Can $Nat$ be an alphabet? No &mdash; it is infinite. We can only ever show a prefix.

```python
nat_prefix = {str(n) for n in range(10)}
print("first 10 naturals as a set :", sorted(nat_prefix))
print("is that an alphabet?        :", is_alphabet(nat_prefix))
print("is Nat itself an alphabet?  : No -- Nat is infinite, alphabets are finite.")
```

The keyboard picture: an alphabet is the whole keyboard, a symbol is one key.

```python
print("Sigma = the keyboard :", sorted(Sigma_bin))
print("a symbol = one key   :", "'0'")
```
