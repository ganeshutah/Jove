## 1. The idea


A **language** is a *possibly empty*, *possibly infinite* **set of strings**, where each
string is **finite**.

The three "possiblys" are the whole definition: emptiness allowed, infinitude allowed,
infinite *strings* never.

Set-builder is how we describe the infinite ones:
$L_{01Rep} = \{(01)^i : i \ge 0\}$, &nbsp; $L_{a\_lt\_b} = \{a^ib^j : i,j \ge 0,\ i<j\}$.

## 2. Definitions

### Two finite languages

```python
EmptyLang = set()
MyLang    = {'', 'aa', 'abc'}
print("EmptyLang :", EmptyLang, " size", len(EmptyLang))
print("MyLang    :", sorted(MyLang), " size", len(MyLang))
```

### Set-builder, as a Python comprehension

The parentheses in $(01)^i$ are **grouping**, not alphabet symbols.

```python
L01Rep_9 = {'01' * i for i in range(5)}
print("L_01Rep (first few) :", sorted(L01Rep_9, key=len))

La_lt_b_9 = {'a'*i + 'b'*j for i in range(10) for j in range(10) if i < j}
print("L_a_lt_b has", len(La_lt_b_9), "strings with i,j < 10")
```

## 3. Tests

Why is $\varepsilon$ **not** in $L_{a\_lt\_b}$? Because $i<j$ forces $j \ge 1$.

```python
print("'' in L_a_lt_b ?", '' in La_lt_b_9)
print("shortest members  :", sorted(La_lt_b_9, key=len)[:4])
assert '' not in La_lt_b_9
print()
print("Change i<j to i<=j and epsilon appears (i=j=0).")
```

Each string is finite; the language is not.

```python
print("longest string we generated :", max(La_lt_b_9, key=len))
print("its length                  :", len(max(La_lt_b_9, key=len)))
print("is the real L_a_lt_b finite? : No -- for any p there is q > p.")
```

Every member really does satisfy the predicate.

```python
def ok(s):
    i = len(s) - len(s.lstrip('a'))
    return s == 'a'*i + 'b'*(len(s)-i) and i < len(s)-i
assert all(ok(s) for s in La_lt_b_9)
print("all", len(La_lt_b_9), "generated strings satisfy  s = a^i b^j with i < j")
```
