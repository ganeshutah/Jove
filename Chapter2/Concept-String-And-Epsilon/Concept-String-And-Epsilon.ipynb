## 1. The idea


A **string** (or **word**) is a *finitely long*, *possibly empty* sequence of symbols,
read left to right. This book never deals with infinite strings.

The **empty string** is $\varepsilon$, written `""` or `''` in Python, and **its length
is 0 by definition**.

"Over an alphabet" is looser than it looks: a string may be regarded as over **any
larger** alphabet &mdash; `"dad"` is a string over $\{a,b,c,d\}$. The only hard requirement
is that every symbol **in** the string is **in** the alphabet.

## 2. Definitions

### $\varepsilon$ and length

```python
eps = ''
print("epsilon repr :", repr(eps), " length :", len(eps))
assert len(eps) == 0
```

### The "over" relation &mdash; one-way containment

```python
def symbols_of(s):
    return set(s)

def is_over(s, Sigma):
    """Every symbol of s must be in Sigma. Sigma may be LARGER."""
    return symbols_of(s) <= Sigma
```

## 3. Tests

`"ateAnApple"` uses eight distinct symbols &mdash; and is over any alphabet containing them.

```python
w = "ateAnApple"
print("symbols in", repr(w), ":", sorted(symbols_of(w)))
print("count :", len(symbols_of(w)))
```

`"dad"` is over $\{a,d\}$ **and** over $\{a,b,c,d\}$ &mdash; the user simply did not press
some keys.

```python
for Sigma in [{'a','d'}, {'a','b','c','d'}, {'a','b'}]:
    print("'dad' over %-18s ? %s" % (sorted(Sigma), is_over('dad', Sigma)))
assert is_over('dad', {'a','b','c','d'}) and not is_over('dad', {'a','b'})
```

$\varepsilon$ is over **every** alphabet &mdash; it has no symbols to violate anything.

```python
print("epsilon over {0,1} :", is_over('', {'0','1'}))
print("epsilon over {}    :", is_over('', set()))
print()
print("Strings are FINITE. This book has no infinite strings anywhere.")
```
