## 1. The idea


Repeated concatenation is exponentiation: $a^2 = aa$, $(ab)^3 = ababab$.

Why is $s^0 = \varepsilon$? For the same reason $x^0 = 1$ for numbers: so that the
recursion $x^n = x \times x^{n-1}$ bottoms out correctly.

**The basis case of a recursive definition is not a free choice.** It is *determined*
by requiring the recursion to be consistent. That argument is replayed for languages
in Concept 12, where it gives a genuinely counter-intuitive result.

## 2. Definitions

### Numeric exponentiation, unfolded

```python
def num_exp(x, n):
    return 1 if n == 0 else x * num_exp(x, n-1)

print("4^3 = 4 * 4^2 = 4 * 4 * 4^1 = 4 * 4 * 4 * 4^0 = 4*4*4*1 =", num_exp(4,3))
```

### String exponentiation, exactly parallel

```python
def str_exp(s, n):
    return '' if n == 0 else s + str_exp(s, n-1)
```

## 3. Tests

The book's examples.

```python
print("a^2        :", repr(str_exp('a',2)))
print("(ab)^3     :", repr(str_exp('ab',3)))
print("(abacaea)^2:", repr(str_exp('abacaea',2)))
assert str_exp('ab',3) == 'ababab' == 'ab'*3
```

Python's `*` agrees, which is the same definition wearing different clothes.

```python
ok = all(str_exp(s,n) == s*n for s in ['a','ab','xyz'] for n in range(6))
print("str_exp agrees with Python's * :", ok)
assert ok
```

**Why the basis is forced.** Set it to anything else and the recursion breaks.

```python
def bad_exp(s, n):
    return 'X' if n == 0 else s + bad_exp(s, n-1)   # wrong basis

print("good a^3 :", repr(str_exp('a',3)))
print("bad  a^3 :", repr(bad_exp('a',3)), " <- the junk basis leaks into every answer")
print()
print("s^0 must be the ONE for concatenation, i.e. epsilon. Not a choice -- a consequence.")
```
