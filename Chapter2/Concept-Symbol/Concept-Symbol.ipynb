## 1. The idea


A **symbol** is whatever you communicate with: an English letter, a whole word, a
musical mark, a smoke signal. Symbols are **primitives in each context** &mdash; you choose
them, you do not define them.

For the rest of the book we require symbols to be **strings of length 1**. In Python a
unit-length string *is* a symbol, and `'a'` and `"a"` mean the same thing.

The real lesson: "symbol" is a **level-of-abstraction choice**, not an absolute.

## 2. Definitions

### Symbols as unit-length strings

```python
sym_a, sym_0 = 'a', '0'
print(sym_a, len(sym_a), "|", sym_0, len(sym_0))
print("'a' == \"a\" :", 'a' == "a")

def is_symbol(x):
    return isinstance(x, str) and len(x) == 1
```

### The same text, two levels of abstraction

Treat a sentence as a sequence of *letters*, or as a sequence of *words*. Both are
legitimate &mdash; they are different choices of what counts as a symbol.

```python
text = "the cat sat"
print("letters as symbols :", [ch for ch in text if ch != ' '][:8], "...")
print("words   as symbols :", text.split())
```

## 3. Tests

Only length-1 strings pass.

```python
for x in ['a', '0', 'ab', '', 5]:
    print("%-5r is a symbol? %s" % (x, is_symbol(x)))
assert is_symbol('a') and not is_symbol('ab') and not is_symbol('')
```

Counting depends on what you called a symbol &mdash; which is the whole point.

```python
print("distinct letter-symbols :", len(set(ch for ch in text if ch != ' ')))
print("distinct word-symbols   :", len(set(text.split())))
```
