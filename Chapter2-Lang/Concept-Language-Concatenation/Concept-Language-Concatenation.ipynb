## 1. The idea


$L_1L_2 = \{xy \;:\; x \in L_1 \wedge y \in L_2\}$.

Read it carefully: **every** $x$ against **every** $y$. Jove's `lcat` is literally that
doubly-nested comprehension.

Because the result is a **set**, duplicates vanish &mdash; so $|L_1L_2|$ can be *less* than
$|L_1| \times |L_2|$. That trap is exercised below.

## 2. Definitions

### Jove's `lcat`

(The book's prose calls it `cat`; the shipped name is **`lcat`**. Appendix B has the
real names &mdash; `l` for language operations, `s` for string ones.)

```python
A = {'ab', 'bc'}
B = {'11', 'ab', '22'}
print("lcat(A,B) :", sorted(lcat(A, B)))
print("size      :", len(lcat(A,B)), "=", len(A), "x", len(B))
```

### The same thing, written out

```python
def cat_by_hand(L1, L2):
    return {x + y for x in L1 for y in L2}
```

## 3. Tests

`lcat` and the hand version agree.

```python
assert cat_by_hand(A,B) == lcat(A,B)
print("agree :", cat_by_hand(A,B) == lcat(A,B))
```

**The collapse trap.** Here $|L_1|\times|L_2| = 4$ but the result has only 3 strings.

```python
C = {'a', 'aa'}
print("C       :", sorted(C))
print("lcat(C,C):", sorted(lcat(C,C)))
print("expected %d pairings, got %d distinct strings" % (len(C)*len(C), len(lcat(C,C))))
assert len(lcat(C,C)) < len(C)*len(C)
print("  'a'+'aa' and 'aa'+'a' both give 'aaa' -- the set absorbs the duplicate.")
```

Concatenation is associative on languages too.

```python
D = {'x'}
assert lcat(lcat(A,B), D) == lcat(A, lcat(B,D))
print("lcat is associative :", lcat(lcat(A,B), D) == lcat(A, lcat(B,D)))
```
