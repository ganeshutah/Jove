## 1. The idea


$$S = \{\,x \ :\ p(x)\,\}$$

"the set of all $x$ satisfying $p$". The predicate $p$ does the work, and two
degenerate choices are worth naming:

* $p = True$ gives the **whole universe** under discussion;
* $p = False$ gives $\emptyset$.

Usually the universe is written in: $\{x \in Nat : x \bmod 2 = 0\}$.

Python's comprehensions are set-builder notation with the colon spelled `for ... if`,
which makes the correspondence easy to check &mdash; and easy to check *mechanically*,
which is what the tests below do.

## 2. Definitions

### Set builder, in mathematics and in Python

```python
U = set(range(20))                       # the universe for these examples

evens   = {x for x in U if x % 2 == 0}   # { x in U : x mod 2 = 0 }
squares = {x * x for x in range(6)}       # { x^2 : x in 0..5 }
print("evens   =", sorted(evens))
print("squares =", sorted(squares))
```

### The two unconditional predicates

```python
TRUE  = lambda x: True
FALSE = lambda x: False

def build(U, p):
    return {x for x in U if p(x)}
```

## 3. Tests

$p = True$ gives the whole universe; $p = False$ gives $\emptyset$.

```python
print("{ x in U : True  } =", sorted(build(U, TRUE))[:8], "...  size", len(build(U, TRUE)))
print("{ x in U : False } =", build(U, FALSE))
assert build(U, TRUE) == U
assert build(U, FALSE) == set()
```

The universe matters &mdash; the same predicate gives different sets.

```python
p = lambda x: x % 3 == 0
for name, uni in [('0..19', set(range(20))), ('0..8', set(range(9))),
                  ('the odds', {x for x in range(20) if x % 2})]:
    print("  { x in %-9s : x mod 3 = 0 } = %s" % (name, sorted(build(uni, p))))
assert build(set(range(9)), p) == {0, 3, 6}
```

A predicate can mention several variables, giving a set of tuples.

```python
pairs = {(a, b) for a in range(4) for b in range(4) if a + b == 3}
print("{ (a,b) in 0..3 x 0..3 : a + b = 3 } =", sorted(pairs))
assert pairs == {(0, 3), (1, 2), (2, 1), (3, 0)}
```

Set builder over **strings** is how languages are written all book long.

```python
from itertools import product
Sigma_star = {''.join(p) for k in range(5) for p in product('01', repeat=k)}
L1 = {w for w in Sigma_star if w.count('0') % 2 == 0}
L2 = {w for w in Sigma_star if w.startswith('1')}
print("|Sigma*| up to length 4 :", len(Sigma_star))
print("L1 (even # of 0s), short :", sorted(w for w in L1 if len(w) <= 2))
print("L2 (starts with 1), short:", sorted(w for w in L2 if len(w) <= 2))
assert '' in L1 and '' not in L2
print("\nNote epsilon: it has zero 0s (even, so in L1) and no first symbol")
print("(so not in L2).  The empty string is where specifications go wrong.")
```
