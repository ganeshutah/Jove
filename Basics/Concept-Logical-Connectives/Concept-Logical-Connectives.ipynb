## 1. The idea


The connectives: **and** $\wedge$, **or** $\vee$, **not** $\neg$, **implies**
$\Rightarrow$, **exclusive-or** $\oplus$.

Two that catch people:

* **$\vee$ is inclusive.** $a \vee b$ is true when both are.
* **$a \Rightarrow b$ is $\neg a \vee b$.** It is **true whenever $a$ is false** &mdash;
  *vacuous truth*. "Every unicorn is purple" is true.

A **predicate** is a Boolean formula with free variables; substituting values gives a
truth value. $True$ and $False$ are the constant predicates, and they are what make
set-builder's two degenerate cases work (Concept 2).

Vacuous truth shows up as real behaviour in Chapter 5: "every block of 3 has exactly
two 1s" is **true** for strings shorter than 3, so they are in the language.

## 2. Definitions

### The connectives, as functions

```python
AND  = lambda a, b: a and b
OR   = lambda a, b: a or b
NOT  = lambda a: not a
IMP  = lambda a, b: (not a) or b
XOR  = lambda a, b: a != b
IFF  = lambda a, b: a == b

def table(name, f, arity=2):
    print("  " + name)
    if arity == 1:
        for a in (False, True): print("     %-6s -> %s" % (a, f(a)))
    else:
        for a in (False, True):
            for b in (False, True):
                print("     %-6s %-6s -> %s" % (a, b, f(a, b)))
```

### A predicate is a formula with free variables

```python
p = lambda x: x % 2 == 0
q = lambda x: x > 5
combined = lambda x: AND(p(x), q(x))
TRUE_p   = lambda x: True
FALSE_p  = lambda x: False
```

## 3. Tests

The truth tables.

```python
for name, f in [("AND", AND), ("OR", OR), ("IMPLIES", IMP), ("XOR", XOR)]:
    table(name, f)
    print()
table("NOT", NOT, arity=1)
```

**$\vee$ is inclusive** &mdash; unlike English 'or'.

```python
print("  OR(True, True) =", OR(True, True), "   XOR(True, True) =", XOR(True, True))
assert OR(True, True) is True and XOR(True, True) is False
print("\n'tea or coffee' in English usually means XOR.  In logic it does not.")
```

**Implication is vacuously true** when the antecedent is false.

```python
for a in (False, True):
    for b in (False, True):
        print("  %-6s => %-6s is %s%s" % (a, b, IMP(a, b),
              "   <- VACUOUS" if not a else ""))
assert IMP(False, False) and IMP(False, True)
print("\nBoth rows with a False antecedent are TRUE.")
```

Which is exactly the behaviour of a universally quantified empty range.

```python
print("  'every block of 3 in \'01\' has exactly two 1s' :",
      all(False for _ in range(0)))
blocks = lambda s: [s[i:i+3] for i in range(len(s) - 2)]
for s in ['', '0', '01', '110', '111']:
    ok = all(b.count('1') == 2 for b in blocks(s))
    print("  %-6r blocks %-12s -> %s%s" % (s, blocks(s), ok,
          "   (vacuous: no blocks)" if not blocks(s) else ""))
assert all(b.count('1') == 2 for b in blocks(''))
print("\nChapter 5, Concept 5 harvests exactly these vacuous cases.")
```

Predicates combine, and the constant predicates are the degenerate ones.

```python
U = set(range(12))
print("  { x : p(x) }        =", sorted(x for x in U if p(x)))
print("  { x : q(x) }        =", sorted(x for x in U if q(x)))
print("  { x : p(x) AND q(x)}=", sorted(x for x in U if combined(x)))
print("  { x : True }        =", sorted(x for x in U if TRUE_p(x))[:5], "...")
print("  { x : False }       =", {x for x in U if FALSE_p(x)})
assert {x for x in U if FALSE_p(x)} == set()
assert {x for x in U if TRUE_p(x)} == U
```
