## 1. The idea


**Proof by contrapositive:** to show $a \Rightarrow b$, show $\neg b \Rightarrow \neg a$.
Licensed by the equivalence of Concept 7.

**Proof by contradiction:** to show $P$, assume $\neg P$ and derive $False$.

They are **the same law**. "Assume $\neg P$, derive $False$" is
$\neg P \Rightarrow False$, whose contrapositive is
$\neg False \Rightarrow \neg\neg P$, that is $True \Rightarrow P$, that is $P$.

The book's teaching choice, stated in Chapter 4, Concept 15: prefer the **direct**
contrapositive form where you can. Assuming $\neg Cond(L)$ and working forwards is
goal-directed from the first line; "assume regular, derive a contradiction" hides the
target.

## 2. Definitions

### The two shapes, as functions

```python
IMP = lambda a, b: (not a) or b
B = [(a, b) for a in (False, True) for b in (False, True)]

def by_contrapositive(a, b): return IMP(not b, not a)
def by_contradiction(P):     return not IMP(not P, False) or P
```

### Worked example: $n^2$ even $\Rightarrow$ $n$ even

```python
def direct_attempt(n):     return (n * n) % 2 == 0 and n % 2 == 0
def contrapositive(n):     # !b => !a :  n odd => n^2 odd
    return IMP(n % 2 == 1, (n * n) % 2 == 1)
```

## 3. Tests

The two forms agree, always.

```python
assert all(IMP(a, b) == by_contrapositive(a, b) for a, b in B)
print("%-8s %-8s %-12s %s" % ("a", "b", "a => b", "!b => !a"))
for a, b in B:
    print("%-8s %-8s %-12s %s" % (a, b, IMP(a, b), by_contrapositive(a, b)))
```

**Contradiction is the contrapositive of the same implication.**

```python
for P in (False, True):
    print("  P = %-6s : (!P => False) = %-6s   and P itself = %s"
          % (P, IMP(not P, False), P))
    assert IMP(not P, False) == P
print("\n!P => False  is literally equivalent to  P.")
print("So the two proof styles are one law wearing two costumes.")
```

The worked example: the contrapositive is the easy direction.

```python
print("claim : n^2 even  =>  n even")
print("direct: from n^2 = 2k, deduce n even -- needs unique factorisation")
print("contra: n odd => n = 2k+1 => n^2 = 4k^2+4k+1, odd.  One line.")
print()
for n in range(12):
    assert contrapositive(n)
print("contrapositive verified for n = 0..11 :", all(contrapositive(n) for n in range(12)))
```

And the original claim follows, checked directly.

```python
bad = [n for n in range(200) if (n * n) % 2 == 0 and n % 2 != 0]
print("counterexamples to 'n^2 even => n even' below 200 :", bad)
assert not bad
```

The book's preference, and why.

```python
print("proof by contradiction : 'suppose L IS regular ... eventually, absurdity'")
print("                         -- you do not know what you are steering toward")
print()
print("direct contrapositive  : 'we show !Cond(L); here is N, here is w, ...'")
print("                         -- goal-directed from line one")
print()
print("Chapter 4, Concept 15 makes this the house style, and it turns the")
print("Pumping Lemma into a recipe rather than a hunt.")
```
