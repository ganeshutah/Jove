## 1. The idea


Regular languages are **closed under reversal** (Chapters 7 and 10 prove it). So:

> if $L_{if}$ were regular, $L_{ifrev}$ would be regular too.

Pump the **reverse** instead. For
$L_{ifrev} = \{c^kb^ja^i : \text{if } i=3 \text{ then } j=k\}$ and $w = c^Nb^Na^3$,
the $|xy|\le N$ clause confines the loop to the **leading $c$s** &mdash; there is nowhere
to hide.

**General tactic: transform, then pump.**

## 2. Definitions

### The reversed language

```python
def in_Lif(s):
    i = len(s) - len(s.lstrip('a')); rest = s[i:]
    j = len(rest) - len(rest.lstrip('b')); k = len(rest) - j
    if rest[j:] != 'c'*k or s != 'a'*i + 'b'*j + 'c'*k: return False
    return (j == k) if i == 3 else True

def in_Lifrev(s):
    return in_Lif(s[::-1])
```

### Pump the reverse

```python
def splits(w, N):
    return [(w[:p], w[p:q], w[q:])
            for p in range(N+1) for q in range(p+1, min(N,len(w))+1)]
```

## 3. Tests

Reversal really does map one language onto the other.

```python
for s in ['aaabbbccc', 'aaabbcc']:
    print("%-12r in L_if=%s   reversed %-12r in L_ifrev=%s"
          % (s, in_Lif(s), s[::-1], in_Lifrev(s[::-1])))
assert in_Lifrev('cccbbbaaa') == in_Lif('aaabbbccc')
```

Now **every** split breaks &mdash; the loop is trapped in the leading $c$s.

```python
N = 4; w = 'c'*N + 'b'*N + 'aaa'
assert in_Lifrev(w)
sp = splits(w, N)
broken = [(x,y,z) for (x,y,z) in sp
          if any(not in_Lifrev(x + y*i + z) for i in range(4))]
print("w =", w)
print("splits: %d,  broken: %d" % (len(sp), len(broken)))
print("distinct y values :", sorted({y for _,y,_ in sp}))
assert len(broken) == len(sp)
print("\nEvery y is c's only -> pumping changes k while j and i stay put -> j != k.")
```

Therefore $L_{if}$ is non-regular after all.

```python
print("L_ifrev is NOT regular  (just proved)")
print("regular languages are closed under reversal")
print("=> if L_if were regular, L_ifrev would be too")
print("=> L_if is NOT regular.  QED")
```
