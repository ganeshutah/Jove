## 1. The idea


Showing $\neg Cond(L)$ shows the pumping condition holds for **no DFA anyone might
propose**. Since every regular language must obey it, that proves **no DFA whatsoever**
exists.

And because you are ruling out *any* machine, you **cannot assume a location for $y$**
&mdash; a different DFA would put the loop elsewhere. Hence: **all splits**.

The good news: a clever choice of $w$ can collapse "all splits" into one uniform case.

## 2. Definitions

### The analogy the book uses

```python
print("Proving  x != x^2  for all x:")
for x in [2, 3, 5]:
    print("   x=%d : %s" % (x, x != x*x))
print("   ... looks proved?")
for x in [0, 1]:
    print("   x=%d : %s   <- counterexample!" % (x, x != x*x))
print("\nCherry-picking the cases that work proves nothing.")
```

### Enumerate every admissible split

```python
def all_splits(w, N):
    return [(w[:i], w[i:j], w[j:])
            for i in range(N+1) for j in range(i+1, min(N, len(w))+1) if j > i]
```

## 3. Tests

For $w = 0^N1^N$, **all** splits confine $y$ to the zeros &mdash; one uniform case.

```python
N = 4; w = '0'*N + '1'*N
sp = all_splits(w, N)
print("number of admissible splits :", len(sp))
print("distinct y values           :", sorted({y for _,y,_ in sp}))
assert all(set(y) <= {'0'} for _,y,_ in sp)
print("\nThe |xy| <= N clause did the work: every y is zeros only.")
```

And each one breaks, so the proof covers them all.

```python
def in_L01(s):
    k = len(s) - len(s.lstrip('0'))
    return s == '0'*k + '1'*(len(s)-k) and k == len(s)-k

broken = [(x,y,z) for (x,y,z) in sp if not in_L01(x + y*0 + z)]
print("splits broken by pumping DOWN (i=0) :", len(broken), "of", len(sp))
assert len(broken) == len(sp)
```

Skip even one split and the argument collapses.

```python
print("If ONE split survived, some DFA might place its loop exactly there,")
print("and that DFA would recognise L. The proof must exclude every placement.")
```
