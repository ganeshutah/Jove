## 1. The idea


A DFA is a directed graph. **Circles are states**; **double circles are final states**;
**labelled arrows are transitions**; an arrow "from the air" marks the **initial state**.

A DFA may have **no** final state, **many**, or **all** of them &mdash; but always
**exactly one initial state**.

**Jove reads the state NAME**: a name starting with `I` is initial, `F` is final, and
`IF` is both. That convention is not cosmetic &mdash; get it wrong and your machine
silently accepts nothing.

## 2. Definitions

### Even number of 0s

We track **parity**, not a count &mdash; so two states suffice.

```python
even0 = md2mc('''DFA
IF : 0 -> A
A  : 0 -> IF
''')
print("states :", sorted(even0["Q"]), " initial:", even0["q0"], " final:", sorted(even0["F"]))
assert even0["q0"] == "IF" and even0["F"] == {"IF"}
```

### The naming rule, demonstrated

A machine with no `F`-named state accepts **nothing**.

```python
oops = md2mc('''DFA
I : 0 -> A
A : 0 -> I
''')
print("final states of 'oops' :", oops["F"], " <- empty!")
assert oops["F"] == set()
```

## 3. Tests

The parity machine works for any length.

```python
for s in ['', '0', '00', '0000', '00000']:
    print("%-7r even # of 0s? %-5s  DFA says %s"
          % (s, len(s) % 2 == 0, accepts_dfa(even0, s)))
assert all(accepts_dfa(even0, '0'*n) == (n % 2 == 0) for n in range(8))
```

The mis-named machine accepts nothing at all &mdash; a silent bug.

```python
print("oops accepts '' ?   ", accepts_dfa(oops, ''))
print("oops accepts '00' ? ", accepts_dfa(oops, '00'))
assert not any(accepts_dfa(oops, '0'*n) for n in range(6))
print("\nAlways check machine['F'] is non-empty when you expect acceptance.")
```
