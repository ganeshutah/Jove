## 1. The idea


The book's stance: **understand the code enough to drive it**, as you understand a car
enough to drive it &mdash; with occasional pushes to go deeper.

* All important functions for a machine type live in **`Def_...`** files:
  `Def_DFA`, `Def_NFA`, `Def_PDA`, `Def_TM`.
* Each lists its functions at the end, and **`help(foo)`** explains any one.
* The notebooks often carry **more detail** than the book's listings.

## 2. Definitions

### What is in `Def_DFA`?

```python
import jove.Def_DFA as DD
fns = [n for n in dir(DD) if not n.startswith('_') and callable(getattr(DD, n))]
print("callables exported by Def_DFA:")
print("  ", ', '.join(sorted(fns)[:18]))
```

## 3. Tests

The docstrings carry the In/Out contract.

```python
for f in [step_dfa, run_dfa, accepts_dfa]:
    doc = (f.__doc__ or '').strip().split('\n')[0]
    print("%-14s %s" % (f.__name__, doc))
assert accepts_dfa.__doc__
```

Which module does a name come from? Useful when a star-import surprises you.

```python
for f in [step_dfa, md2mc, dotObj_dfa]:
    print("%-12s from %s" % (f.__name__, f.__module__))
```
