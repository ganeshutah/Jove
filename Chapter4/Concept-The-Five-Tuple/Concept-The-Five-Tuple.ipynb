## 1. The idea


A DFA is $(Q,\Sigma,\delta,q_0,F)$:

* $Q$ &mdash; a **finite non-empty** set of states,
* $\Sigma$ &mdash; a **finite non-empty** alphabet,
* $\delta: Q\times\Sigma \to Q$ &mdash; a **total** transition function,
* $q_0 \in Q$ &mdash; the initial state,
* $F \subseteq Q$ &mdash; the (possibly empty) final states.

Jove's dictionary has exactly those five keys, so the mathematics and the data
structure are the same shape.

## 2. Definitions

### The five keys

```python
D = md2mc('''DFA
I : 0 -> I
I : 1 -> F
F : 0 -> F
F : 1 -> I
''')
for k in ['Q', 'Sigma', 'Delta', 'q0', 'F']:
    v = D[k]
    print("%-6s : %s" % (k, sorted(v) if isinstance(v, (set, list)) else v))
```

### $\delta$ as a state table

```python
def state_table(D):
    rows = [("State", "Input", "Next")]
    for (q, c), nxt in sorted(D["Delta"].items()):
        rows.append((q, c, nxt))
    for r in rows:
        print("%-8s %-6s %s" % r)
```

## 3. Tests

$\delta$ is **total**: one entry per (state, symbol) pair.

```python
expected = len(D["Q"]) * len(D["Sigma"])
print("|Q| x |Sigma| =", expected, "   entries in Delta:", len(D["Delta"]))
assert len(D["Delta"]) == expected
```

The state table, which is the third way to present the same $\delta$.

```python
state_table(D)
```

Building the same machine by hand with `mk_dfa` gives the same language.

```python
D2 = mk_dfa({'I','F'}, {'0','1'},
            {('I','0'):'I', ('I','1'):'F', ('F','0'):'F', ('F','1'):'I'},
            'I', {'F'})
print("same language as the markdown version?", langeq_dfa(D, D2))
assert langeq_dfa(D, D2)
```
