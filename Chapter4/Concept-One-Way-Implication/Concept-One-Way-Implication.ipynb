## 1. The idea


The lemma has the form $Regular(L) \Rightarrow Cond(L)$.

> **Never use it to prove something *is* regular.** Use it only to **disprove**.

The instrument is the **contrapositive**: $\neg Cond(L) \Rightarrow \neg Reg(L)$.

The book also makes a teaching choice: rather than proof by contradiction, start
**directly** by assuming $Cond(L)$ is false. That is goal-directed from the first line
and turns into a reliable recipe.

## 2. Definitions

### Satisfying $Cond$ does not make a language regular

```python
# A NON-regular language whose strings nonetheless pump on some splits.
def in_L(s):
    return s.count('0') == s.count('1') and s == '0'*s.count('0') + '1'*s.count('1')

print("0^n 1^n members :", [s for s in ['', '01', '0011', '000111'] if in_L(s)])
```

### The logical shapes, side by side

```python
def implies(a, b):  return (not a) or b
def contrapositive(a, b): return implies(not b, not a)
```

## 3. Tests

$P \Rightarrow Q$ and $\neg Q \Rightarrow \neg P$ are the same statement.

```python
rows = [(a,b) for a in (True,False) for b in (True,False)]
print("%-6s %-6s %-10s %s" % ("Reg", "Cond", "Reg=>Cond", "!Cond=>!Reg"))
for a,b in rows:
    print("%-6s %-6s %-10s %s" % (a, b, implies(a,b), contrapositive(a,b)))
assert all(implies(a,b) == contrapositive(a,b) for a,b in rows)
```

The direction that is **invalid**: $Cond$ true tells you nothing.

```python
print("Reg => Cond  is given.")
print("Cond true, Reg? ->", "UNKNOWN -- the implication does not run backwards")
print()
print("So: satisfying the pumping condition NEVER proves regularity.")
print("To prove a language regular, BUILD A DFA.")
```

The right way to prove regularity: exhibit a machine.

```python
D = md2mc('''DFA
IF : 0 -> IF
IF : 1 -> F
F  : 0 | 1 -> F
''')
print("L = 'strings of 0s, optionally then anything' -- regular, here is the DFA:")
print("  states:", sorted(D["Q"]))
assert accepts_dfa(D, '000') and accepts_dfa(D, '0001')
print("\nA machine is a PROOF of regularity. The Pumping Lemma never is.")
```
