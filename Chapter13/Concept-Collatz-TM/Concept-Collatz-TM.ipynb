## 1. The idea


The **Collatz** rule: from $n$, go to $n/2$ if even and $3n+1$ if odd. Every starting
value anyone has tried reaches 1. **Nobody knows whether all do.**

Write a machine that applies the rule and halts on reaching 1. Whether it halts on
*every* input is an **open mathematical problem** &mdash; not a gap in our tooling, but a
genuinely unanswered question.

That is what makes the halting problem concrete. It is not that deciding halting is
*hard to implement*; it is that answering it for a machine this small can require
settling a conjecture that has resisted eighty years of effort.

And this is only one machine. A halting decider would have to settle *every* such
question.

## 2. Definitions

### The Collatz step, and the trajectory

```python
def collatz_step(n):
    return n // 2 if n % 2 == 0 else 3 * n + 1

def trajectory(n, cap=1000):
    out = [n]
    while n != 1 and len(out) < cap:
        n = collatz_step(n); out.append(n)
    return out
```

### A TM-shaped rendering: unary tape, one rule per parity

```python
def collatz_tm_steps(n, fuel=10000):
    # what a TM applying the rule on a unary tape would do, counted in
    # rule applications rather than head movements
    steps = 0
    while n != 1 and steps < fuel:
        n = collatz_step(n); steps += 1
    return (n == 1), steps
```

## 3. Tests

Every starting value tried reaches 1.

```python
for n in [6, 7, 27]:
    t = trajectory(n)
    print("  n=%-3d reaches 1 in %3d steps, peak %d" % (n, len(t) - 1, max(t)))
    assert t[-1] == 1
```

Including some that take a remarkably long time.

```python
worst = max(range(1, 2000), key=lambda n: len(trajectory(n)))
t = trajectory(worst)
print("  longest start below 2000 : n=%d, %d steps, peak %d"
      % (worst, len(t) - 1, max(t)))
assert t[-1] == 1
```

The machine halts on everything we can test &mdash; which proves nothing.

```python
allhalt = all(collatz_tm_steps(n)[0] for n in range(1, 3000))
print("all starts below 3000 halt? ", allhalt)
assert allhalt
print()
print("Tested: 3000 values.  Claimed: all values.  Gap: infinite.")
```

**Halting here is an open problem.**

```python
print("Does this machine halt on every input?")
print("   -- unknown since 1937")
print("   -- verified by computer past 2^68")
print("   -- no proof, no counterexample")
print()
print("A halting decider would answer this question by running.  That alone")
print("should make you suspect no such decider exists.")
```

And this is one machine among infinitely many such questions.

```python
OPEN = ["Collatz (3x+1)", "Goldbach's conjecture as a search loop",
        "twin primes as a search loop", "the Riemann hypothesis, suitably encoded"]
for o in OPEN:
    print("  * a machine whose halting is equivalent to:", o)
print("\nEach is a TM of modest size.  A decider would settle them all.")
```
