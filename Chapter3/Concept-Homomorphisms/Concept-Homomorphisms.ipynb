## 1. The idea


$h : \Sigma^* \to \Gamma^*$ is a **string homomorphism** if

* $h(\varepsilon) = \varepsilon$, and
* $h(xy) = h(x)h(y)$ &mdash; pull the string apart anywhere, map the pieces, rejoin.

The **language homomorphism** is $h(L) = \{h(x) : x \in L\}$.

Jove takes `f` as a **lambda from characters to characters**, which is exactly the
condition that guarantees no conflicts (Concept 11).

## 2. Definitions

### The Caesar shift from the book

```python
hm = lambda x: chr((ord(x)+1) % 256)
print("shomo('Hello there', hm) :", shomo('Hello there', hm))
print("shomo('abcd', hm)        :", shomo('abcd', hm))
```

### Two more, with very different purposes

```python
tap    = lambda x: 't'                        # deliberate forgetting
rename = lambda x: 'a' if x == '0' else 'b'   # alphabet renaming
```

## 3. Tests

The two axioms, checked.

```python
print("h(epsilon) == epsilon :", shomo('', hm) == '')
x, y = 'Hell', 'o there'
print("h(x)      :", shomo(x, hm))
print("h(y)      :", shomo(y, hm))
print("h(x)h(y)  :", shomo(x,hm) + shomo(y,hm))
print("h(xy)     :", shomo(x+y, hm))
assert shomo(x+y, hm) == shomo(x,hm) + shomo(y,hm)
assert shomo('', hm) == ''
```

It holds for **every** way of splitting the string &mdash; that is the force of the axiom.

```python
w = 'Hello there'
ok = all(shomo(w,hm) == shomo(w[:i],hm) + shomo(w[i:],hm) for i in range(len(w)+1))
print("h(xy)=h(x)h(y) for all %d splits of %r : %s" % (len(w)+1, w, ok))
assert ok
```

`lhomo` lifts it to a language. The "tap" map **forgets** what was typed.

```python
L = {'01','10','0011'}
print("lhomo(L, rename) :", sorted(lhomo(L, rename)))
print("lhomo(L, tap)    :", sorted(lhomo(L, tap)), " <- only the LENGTHS survive")
assert lhomo(L, tap) == {'tt','tttt'}
print()
print("That forgetting trick is how Chapter 8 reads off the LENGTHS of a language.")
```
