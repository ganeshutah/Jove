## 1. The idea


Key facts exercised here:

* Exactly **two** languages have a finite star: $\emptyset$ and $\{\varepsilon\}$.
* $L_7=\{0^i1^j\} = \{0\}^*\{1\}^*$ &mdash; yes, because the exponents were already free.
* $L_{P2}=\{waw^R : a \in \{\varepsilon\}\cup\Sigma\}$ is **all** palindromes; the
  $\varepsilon$-or-symbol middle is the trick.
* $L_O^* = \{0\}^*$ but $L_E^* \ne \{0\}^*$.

To show $A=B$ for languages: show $A \subseteq B$ and $B \subseteq A$.

## 2. Definitions

### The palindrome family

```python
B = 5
Sig = {'0','1'}
Sst = lstar(Sig, B)

LP1 = {w + w[::-1]             for w in Sst}                  # even palindromes
LP3 = {w + a + w[::-1]         for w in Sst for a in Sig}      # odd palindromes
LP2 = {w + a + w[::-1]         for w in Sst for a in list(Sig) + ['']}
Lww = {w + w                   for w in Sst}                  # a COPY, not a reversal
```

### Even- and odd-length zero strings

```python
LE = {'0'*(2*i)     for i in range(B+1)}
LO = {'0'*(2*i + 1) for i in range(B+1)}
```

## 3. Tests

Only $L_{P2}$ is **all** palindromes &mdash; $L_{P1}$ misses the odd ones.

```python
def is_pal(s): return s == s[::-1]
short = {s for s in Sst if len(s) <= 4}
pals  = {s for s in short if is_pal(s)}
print("all palindromes up to len 4 :", len(pals))
print("caught by LP1 (even only)   :", len(pals & LP1))
print("caught by LP2 (eps or sym)  :", len(pals & LP2))
assert pals <= LP2 and not (pals <= LP1)
print("\n'010' is an ODD palindrome -> in LP2, not in LP1 :", '010' in LP2, '010' in LP1)
```

$ww^R$ versus $ww$ &mdash; the Chapter 1 boundary, in data.

```python
print("'0110' is w w^R (palindrome) :", '0110' in LP1, " is it w w? ", '0110' in Lww)
print("'0101' is w w   (a copy)     :", '0101' in Lww, " is it w w^R?", '0101' in LP1)
print()
print("w w^R : CONTEXT-FREE.   w w : NOT context-free. Reversal is cheap, copying is not.")
```

$L_O^* = \{0\}^*$ but $L_E^* \ne \{0\}^*$.

```python
zeros = lstar({'0'}, 8)
LOstar = lstar(LO, 4); LEstar = lstar(LE, 4)
print("odd lengths generate everything? :", lissubset({s for s in zeros if len(s)<=6},
                                                      {s for s in LOstar if len(s)<=6}))
print("'0' in LE* ? :", '0' in LEstar, "  <- even-only can never make an odd length")
assert '0' not in LEstar
```

$L_OL_O$ is **not** $L_E$: it misses $\varepsilon$.

```python
LOLO = lcat(LO, LO)
print("in LE but not in LO.LO :", sorted(lminus({s for s in LE if len(s)<=6}, LOLO)))
assert '' in lminus(LE, LOLO)
print("  -> exactly {epsilon}: odd + odd is even, but never zero-length.")
```
