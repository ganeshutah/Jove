## 1. The idea


**Definition 1.** $L^* = L^0 \cup L^1 \cup L^2 \cup \ldots$

**Definition 2.** $L^* = \bigcup_{i=0}^{\infty} L^i$ &nbsp; (the same, compactly)

**Definition 3.** $L^* = \{x \;:\; \exists k \in Nat,\ x \in L^k\}$

Definition 3 turns membership into a **single existential claim over $k$**, which is
why the $L^{**} = L^*$ proof runs on it rather than on unions.

Star is repeated selection from $L$ followed by concatenation &mdash; and **the selection is
re-made at every step**.

## 2. Definitions

### Definition 1, as a bounded union

```python
def star_def1(L, n):
    out = lunit()
    for i in range(1, n+1):
        out = lunion(out, lexp(L, i))
    return out
```

### Definition 3, as a membership test

"Is there some $k$ with $x \in L^k$?"

```python
def in_star_def3(x, L, kmax=8):
    return any(x in lexp(L, k) for k in range(kmax+1))
```

## 3. Tests

All three agree on a bounded window.

```python
L = {'ab', 'b'}
d1 = star_def1(L, 4)
d2 = lstar(L, 4)
print("Definition 1 == Jove's lstar :", d1 == d2)
assert d1 == d2
```

Definition 3 agrees with membership in the union.

```python
for x in ['', 'ab', 'b', 'abb', 'abab', 'a', 'ba']:
    print("%-6r in star? def3=%-5s  in lstar(L,6)=%s"
          % (x, in_star_def3(x, L), x in lstar(L,6)))
assert all(in_star_def3(x,L) == (x in lstar(L,6))
           for x in ['','ab','b','abb','abab','a','ba'])
```

**The selection is re-made at each step.** That is what makes $ab\,b\,ab$ a member.

```python
print("'ababb' in star? :", in_star_def3('ababb', L))
print("built as: ab + ab + b   <- a different choice at each of the three steps")
```
