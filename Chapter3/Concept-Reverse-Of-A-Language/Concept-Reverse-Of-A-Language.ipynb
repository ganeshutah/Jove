## 1. The idea


$rev(L) = \{rev(s) : s \in L\}$ &mdash; lift a string operation to a language by mapping it
over the set. The same pattern as homomorphism (Concept 10).

Python has no built-in string reverse; the idiom is the backward slice `s[::-1]`.

**Naming**: the book's prose says `revs`/`revl`; **Jove ships `srev`/`lrev`**.
Appendix B has the real names.

## 2. Definitions

### The two levels

```python
print("srev('ab')          :", srev('ab'))
print("lrev({'ab','bc'})   :", sorted(lrev({'ab','bc'})))
print("the idiom           :", 'ab'[::-1])
```

### Reversal is an *anti*-homomorphism

```python
def check_homo_law(x, y):
    return srev(x+y) == srev(x)+srev(y), srev(x+y) == srev(y)+srev(x)
```

## 3. Tests

Reversal is an involution: doing it twice returns you home.

```python
L = {'ab','bc','xyz'}
assert lrev(lrev(L)) == L
print("lrev(lrev(L)) == L :", lrev(lrev(L)) == L)
```

A palindrome language is its own reverse.

```python
pals = {w + w[::-1] for w in lstar({'0','1'}, 3)}
print("lrev(pals) == pals :", lrev(pals) == pals)
assert lrev(pals) == pals
```

**Is reversal a homomorphism?** No &mdash; it reverses the *order of the pieces*.

```python
for x, y in [('ab','cd'), ('a','bc')]:
    same, swapped = check_homo_law(x,y)
    print("x=%-4s y=%-4s   rev(xy)==rev(x)rev(y)? %-5s   ==rev(y)rev(x)? %s"
          % (x, y, same, swapped))
assert not check_homo_law('ab','cd')[0]
assert check_homo_law('ab','cd')[1]
print()
print("h(xy) = h(y)h(x), not h(x)h(y). So reversal is an ANTI-homomorphism.")
```
