## 1. The idea


Languages are sets, so complement is subtraction from the universe:

$$\overline{L} = \{x \;:\; x \in (\Sigma^* - L)\}$$

&mdash; "all strings typeable on a keyboard with keys in $\Sigma$ that do not fall within $L$".

`lcomplem(L, Sigma, n)` needs **three** arguments: `Sigma` because complement is
meaningless without a universe, and `n` because that universe is infinite.

## 2. Definitions

### Complement, bounded

```python
Sigma = {'0','1'}
L = {'01', '10'}
print("lcomplem(L, Sigma, 2) :", sorted(lcomplem(L, Sigma, 2), key=lambda s:(len(s),s)))
```

### The same thing at the machine level, where it is **exact**

```python
ends1 = md2mc('''DFA
I : 0 -> I
I : 1 -> F
F : 0 -> I
F : 1 -> F
''')
comp_ends1 = comp_dfa(ends1)
print("original final states  :", sorted(ends1["F"]))
print("complement final states:", sorted(comp_ends1["F"]))
```

## 3. Tests

The complement depends on the **alphabet**.

```python
for Sig in [{'0','1'}, {'0','1','2'}]:
    comp = lcomplem(L, Sig, 2)
    print("alphabet %-12s -> %2d strings in the complement" % (sorted(Sig), len(comp)))
print()
print("Same L, different complements. That is why lcomplem needs Sigma.")
```

And on the **bound** &mdash; it never converges.

```python
for n in [1,2,3,4]:
    print("bound %d -> %2d strings" % (n, len(lcomplem(L, Sigma, n))))
print("The true complement is infinite; we only ever see a window.")
```

The machine version is **exact**: every string is classified, one way or the other.

```python
for s in ['', '0', '1', '01', '10', '11']:
    a, b = accepts_dfa(ends1, s), accepts_dfa(comp_ends1, s)
    print("%-4r original=%-5s complement=%-5s  exactly one? %s" % (s, a, b, a != b))
assert all(accepts_dfa(ends1,s) != accepts_dfa(comp_ends1,s) for s in lstar(Sigma,4))
print("\nFor all 31 strings of length <= 4: exactly one machine accepts. No approximation.")
```
