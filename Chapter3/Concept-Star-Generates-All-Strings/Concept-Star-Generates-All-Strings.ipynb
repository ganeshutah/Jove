## 1. The idea


`lstar(Sigma,n)` gives all strings over $\Sigma$ of length $\le n$; the real star is
$n \to \infty$. So $\{0,1\}^*$ contains **all finite strings** over $0,1$.

Take $L=\{10,01,100\}$ and $M = L^*$. Every string in $M$ is **finite**, being a
concatenation of finitely many members. Yet $M$ itself is **infinite**, because for
every $s \in M$ there is a longer $s' \in M$.

And $N = M^* = L^{**}$ still has only finitely long strings &mdash; which foreshadows
Concept 12.

## 2. Definitions

### Watch `lstar` fill up $\Sigma^*$ level by level

```python
Sig01 = {'0','1'}
for n in range(4):
    print("lstar(Sig01,%d) : %s" % (n, sorted(lstar(Sig01,n), key=lambda s:(len(s),s))))
```

### $M = L^*$ for $L=\{10,01,100\}$

```python
L = {'10','01','100'}
M = lstar(L, 3)
print("|M| at bound 3 :", len(M))
print("longest string :", max(M, key=len), "of length", max(len(s) for s in M))
```

## 3. Tests

Every string is finite; the set keeps growing.

```python
for n in range(5):
    Mn = lstar(L, n)
    print("bound %d : %4d strings, longest %2d symbols" % (n, len(Mn), max(len(s) for s in Mn)))
print()
print("The COUNT grows without limit. The LENGTH of any one string is always finite.")
```

For every string there is a longer one &mdash; so $M$ is infinite.

```python
M4 = lstar(L,4)
s  = max(lstar(L,3), key=len)
longer = [t for t in M4 if len(t) > len(s)]
print("a longest string at bound 3 :", s, "(len %d)" % len(s))
print("longer ones exist at bound 4 :", len(longer) > 0, " e.g.", longer[0])
assert longer
```

$N = M^*$ also has only finite strings &mdash; foreshadowing $L^{**} = L^*$.

```python
N = lstar(M, 1)
print("|N| :", len(N), " longest:", max(len(t) for t in N))
print("every string in N finite? :", all(len(t) < float('inf') for t in N))
print()
print("Starring a starred language adds no infinitely-long strings. See Concept 12.")
```
