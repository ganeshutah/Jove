## 1. The idea


**Lexicographic** order is dictionary order: `pig` before `poblano`, `pig` after `pago`.

Applied to an infinite set it **starves**: listing $\{a,b\}^*$ lexicographically gives
$\varepsilon, a, aa, aaa, \ldots$ and **never reaches $b$**.

So we need a stronger notion. **Enumeration** = listing *with the guarantee of reaching
every string after a finite amount of other strings*. That finite-waiting criterion is
what "countable" means for us.

## 2. Definitions

### Lexicographic listing of $\{a,b\}^*$

```python
def lex_listing(n):
    """Sort strings in pure dictionary order -- and watch what comes first."""
    pool = sorted(lstar({'a','b'}, n))
    return pool
```

### An enumeration, grouped by length

```python
def enumeration(n):
    return sorted(lstar({'a','b'}, n), key=lambda s: (len(s), s))
```

## 3. Tests

Lexicographic order front-loads all the `a`s.

```python
lex = lex_listing(4)
print("first 12 lexicographically :", lex[:12])
print("position of 'b' in that list :", lex.index('b'))
print("strings before 'b'           :", lex.index('b'))
print()
print("Raise the bound and 'b' slides further back -- in the infinite set it NEVER arrives.")
```

Growing the bound pushes `b` ever later &mdash; the starvation, in numbers.

```python
for n in range(2, 7):
    l = sorted(lstar({'a','b'}, n))
    print("bound %d : 'b' is at index %3d of %3d" % (n, l.index('b'), len(l)))
print("\nIn the limit the index is infinite. 'b' is never listed.")
```

The enumeration reaches everything after finitely many others.

```python
e = enumeration(3)
print("enumeration :", e[:15])
print("'b' at index :", e.index('b'), " <- reached almost immediately")
assert e.index('b') < 5
print()
print("ENUMERATION = every string arrives after only FINITELY many others.")
```
