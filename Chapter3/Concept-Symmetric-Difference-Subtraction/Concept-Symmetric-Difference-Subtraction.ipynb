## 1. The idea


$L_1 - L_2 = L_1 \cap \overline{L_2} = \{x : x \in L_1 \wedge x \notin L_2\}$

$(L_1-L_2) \cup (L_2-L_1)$ is the **symmetric difference**.

Its diagnostic use is the point: it is empty **exactly when the two languages are
equal**, and its members are the strings on which they disagree.

## 2. Definitions

### Subtraction and symmetric difference

```python
A = {'a','ab','b'}
B = {'b','ba'}
print("lminus(A,B)   :", sorted(lminus(A,B)))
print("lminus(B,A)   :", sorted(lminus(B,A)))
print("lsymdiff(A,B) :", sorted(lsymdiff(A,B)))
```

### Building $a^nb^nc^n$ by intersection

Two simpler languages intersect to the canonical **non**-context-free one.

```python
# The bound must be big enough to hold 'aabbcc' (6 symbols), so we take
# strings up to length 6 on the left, and at most 2 of each letter on the right.
eqabc   = {s for s in lstar({'a','b','c'}, 6)
           if s.count('a') == s.count('b') == s.count('c')}
ordered = lcat(lcat(lstar({'a'},2), lstar({'b'},2)), lstar({'c'},2))
print("eqabc holds strings up to length", max(len(s) for s in eqabc))
```

## 3. Tests

Equality test: empty symmetric difference.

```python
X = lstar({'0'}, 4)
Y = {'0'*i for i in range(5)}
print("lsymdiff(X,Y) :", lsymdiff(X,Y))
print("X == Y ?      :", lsymdiff(X,Y) == set())
assert lsymdiff(X,Y) == set()
```

When they differ, the difference **names the witnesses**.

```python
Z = {'0'*i for i in range(4)}          # missing '0000'
d = lsymdiff(X, Z)
print("lsymdiff(X,Z) :", sorted(d))
print("  -> those strings are exactly where the two definitions disagree.")
assert d
```

Intersection builds $a^nb^nc^n$ from two easier pieces.

```python
anbncn = lint(eqabc, ordered)
print("a^n b^n c^n (bounded) :", sorted(anbncn, key=len))
assert 'abc' in anbncn and 'aabbcc' in anbncn and 'acb' not in anbncn
print()
print("Each piece alone is easy. Their INTERSECTION is not even context-free (Ch 11).")
```
