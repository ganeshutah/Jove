## 1. The idea


**Numeric order**: exhaust each **length group** before moving on, and use
lexicographic order *within* a group.

So `pig` still comes before `poblano`, but `pago` now comes **after** `pig` &mdash; all
length-3 strings are listed before any length-4 one.

Every string arrives after finitely many others (each group has $|\Sigma|^n$ members),
so numeric order **is** an enumeration. It is sometimes called *shortlex*.

## 2. Definitions

### The two-level key

```python
def numeric_order(strings):
    return sorted(strings, key=lambda s: (len(s), s))
```

### The `pig` / `pago` / `poblano` comparison

```python
words = ['pig', 'poblano', 'pago']
```

## 3. Tests

Lexicographic vs numeric, on the book's three words.

```python
print("lexicographic :", sorted(words))
print("numeric       :", numeric_order(words))
print()
print("'pago' moves AFTER 'pig' -- length 4 waits for all of length 3.")
assert numeric_order(words).index('pig') < numeric_order(words).index('pago')
```

Group sizes are $|\Sigma|^n$, so every group is finite.

```python
for n in range(5):
    grp = [s for s in lstar({'0','1'}, n) if len(s) == n]
    print("length %d : %2d strings  (2^%d = %d)" % (n, len(grp), n, 2**n))
assert all(len([s for s in lstar({'0','1'},n) if len(s)==n]) == 2**n for n in range(5))
```

Jove's `nthnumeric` produces exactly this order.

```python
mine = numeric_order(lstar({'0','1'}, 3))
jove = [nthnumeric(i, ['0','1']) for i in range(len(mine))]
print("mine :", mine[:10])
print("jove :", jove[:10])
assert mine == jove
print("\nMy two-level sort and Jove's nthnumeric agree exactly.")
```
