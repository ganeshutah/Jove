## 1. The idea


Why prove things about a machine nobody builds?

Because an **impossibility** proof must quantify over *all* procedures, and you cannot
quantify over "all laptops" &mdash; the notion is not mathematical. Turing machines give
you a **formal object to induct over, diagonalize against, and reduce to**.

Two consequences students often resist:

* **speed cannot help.** Undecidability is not about time. A machine a trillion times
  faster still cannot decide the halting problem; there is no finite time bound to
  beat.
* **more memory cannot help.** A TM already has unbounded tape. Adding resources to an
  already-unbounded model changes nothing.

The formal model is a **vehicle for proof**, not an engineering proposal.

## 2. Definitions

### Speed does not change what is decidable

```python
Looper = md2mc('''TM
I : 0 ; 0 , R -> I
I : . ; . , R -> I      !! runs right forever
''')
Halter = md2mc('''TM
I : 0 ; 0 , R -> I
I : . ; . , S -> F
''')

# --- thin wrappers over Jove's TM runner --------------------------------
# run_tm(T, tape, fuel) returns (truncated-paths, haltList).  A TM HALTS
# when no transition applies, and ACCEPTS if it halts in a final state.
# So an accepting state must have NO outgoing transitions, or the machine
# will run on past it.
def tm_accepts(T, tape, fuel=200):
    trunc, halts = run_tm(T, tape if tape != '' else '.', fuel, chatty=False)
    return any(cfg[0] in T["F"] for cfg, _ in halts)

def tm_halts(T, tape, fuel=200):
    trunc, halts = run_tm(T, tape if tape != '' else '.', fuel, chatty=False)
    return len(halts) > 0

def tm_tape(T, tape, fuel=200):
    trunc, halts = run_tm(T, tape if tape != '' else '.', fuel, chatty=False)
    return [cfg[2].rstrip('.') for cfg, _ in halts]
```

### Simulating 'a faster machine' by giving more fuel

```python
def observe(T, tape, fuels):
    return [(f, tm_halts(T, tape, fuel=f)) for f in fuels]
```

## 3. Tests

More 'speed' (fuel) never converts a non-halting run into a halting one.

```python
print("Looper :", observe(Looper, '000', [10, 50, 200, 1000]))
print("Halter :", observe(Halter, '000', [2, 4, 10, 50]))
assert not any(h for _, h in observe(Looper, '000', [10, 50, 200, 1000]))
assert tm_halts(Halter, '000', fuel=50)
```

You can always be told 'not yet', and never 'never'.

```python
for f in [10, 100, 1000]:
    print("  after %4d steps, Looper has not halted. Will it ever? unknown from here."
          % f)
print("\nNo finite observation distinguishes 'has not halted yet' from 'never halts'.")
```

Unbounded memory is already in the model, so adding more changes nothing.

```python
print("tape           : already unbounded")
print("steps          : already unbounded")
print("alphabet size  : provably reducible to 2 symbols")
print("number of tapes: provably reducible to 1")
print()
print("Every 'upgrade' you can name is already absorbed by the model.")
```

What the formal model buys: objects you can do mathematics to.

```python
TECHNIQUES = [("diagonalization", "build a machine that differs from every machine in a list"),
              ("reduction",       "if A were decidable, so would B be -- but B is not"),
              ("simulation",      "one machine runs another, so classes collapse"),
              ("counting",        "there are countably many TMs and uncountably many languages")]
for t, w in TECHNIQUES: print("  %-16s %s" % (t, w))
print("\nNone of these can be applied to 'a laptop'.")
```
